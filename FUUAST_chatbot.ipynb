```python
!pip install -U google-genai

from google import genai
from google.colab import userdata

client = genai.Client(api_key=userdata.get("GEMINI_API_KEY"))

MODEL = "gemini-3.6-flash"

academic_data = {
    "degree_levels": [
        "Bachelor's",
        "Master's",
        "PhD"
    ],

    "academic_areas": [
        "Computer Science",
        "Software Engineering",
        "Artificial Intelligence",
        "Business Administration",
        "Mathematics"
    ],

    "curriculum_information": [
        "Courses and subjects",
        "Course syllabi",
        "Credit hours",
        "Degree requirements",
        "Majors",
        "Specializations",
        "Areas of study"
    ]
}


def university_agent(user_input):

    prompt = f"""
You are an AI Agent of Federal Urdu University of Arts, Science and Technology (FUUAST).

Academic Programs & Curriculum: Degrees, courses, credit hours, majors, and areas of study.

Your main purpose is to provide information about the academic programs and curriculum of FUUAST.

You have a local Python knowledge base containing academic information.

LOCAL ACADEMIC KNOWLEDGE:

{academic_data}

IMPORTANT INSTRUCTIONS:

1. Use the local academic knowledge when answering questions about programs, courses, credit hours, majors, specializations, and areas of study.

2. Use your general knowledge to provide additional useful information when appropriate.

3. Do not invent specific university information that is not available in the local knowledge or your reliable general knowledge.

4. If specific information is not available, clearly state that the information is not available in the current knowledge base.

5. For current or official university information, mention that it should be verified from the official FUUAST source.

6. Keep answers simple, clear, concise, and helpful.

7. If the question is unrelated to academic programs or FUUAST, politely explain that you are a FUUAST Academic Information Agent.

User Question:
{user_input}

Agent Response:
"""

    response = client.models.generate_content(
        model=MODEL,
        contents=prompt
    )

    return response.text


print("==============================================")
print("FUUAST Academic Information Agent")
print("==============================================")
print()
print("Academic Programs & Curriculum: Degrees, courses, credit hours, majors, and areas of study.")
print()
print("Ask your academic question below.")
print("Type 'exit' to stop.")
print("==============================================\n")


while True:

    user = input("You: ")

    if user.lower() == "exit":
        print("Goodbye!")
        break

    try:
        answer = university_agent(user)
        print("\nAgent:", answer, "\n")

    except Exception as e:
        print("Error:", e)
```
