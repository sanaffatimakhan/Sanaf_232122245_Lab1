```python
!pip install -U google-genai

from google import genai
from google.colab import userdata

client = genai.Client(api_key=userdata.get("GEMINI_API_KEY"))

MODEL = "gemini-3.6-flash"

teachers = {
    "Dr. Uzma Afzal": [
        "Software Engineering",
        "Software Project Management",
        "Software Process Models"
    ],

    "Dr. Muhammad Ali": [
        "Operating Systems",
        "Computer Networks",
        "Computer Architecture"
    ],

    "Dr. Sana Khan": [
        "Database Systems",
        "Database Management",
        "Data Warehousing"
    ],

    "Dr. Ahmed Raza": [
        "Artificial Intelligence",
        "Machine Learning",
        "Data Science"
    ],

    "Dr. Fatima Noor": [
        "Multivariate Calculus",
        "Linear Algebra",
        "Discrete Mathematics"
    ],

    "Dr. Hassan Ahmed": [
        "Data Structures and Algorithms",
        "Programming Fundamentals",
        "Object Oriented Programming"
    ]
}

teacher_information = ""

for teacher, subjects in teachers.items():
    teacher_information += f"{teacher}: {', '.join(subjects)}\n"


def university_agent(user_input):

    prompt = f"""
You are an AI Agent of Federal Urdu University of Arts, Science and Technology (FUUAST).

Your main purpose is to help users with information about:
- Which teacher teaches which subject
- Which subjects are taught by a particular teacher

You have a local Python knowledge base containing specific teacher-subject information.

LOCAL TEACHER KNOWLEDGE:
{teacher_information}

IMPORTANT INSTRUCTIONS:

1. For teacher-subject questions, use the provided local teacher knowledge.
2. Do not invent or change teacher-subject assignments.
3. If a teacher or subject is not present in the local knowledge, clearly say that this specific information is not available in the provided teacher database.
4. For common/general information about Federal Urdu University, such as its city, general campus location, university background, or other commonly known information, use your existing general knowledge.
5. You do not need the user to provide information that you already know from your general knowledge.
6. If the user asks a question related to the university but outside teacher-subject information, answer it using your general knowledge when possible.
7. If the question is completely unrelated to the university, politely explain that you are a Federal Urdu University AI Agent.
8. Keep answers clear, concise, and helpful.

User question:
{user_input}

Agent response:
"""

    response = client.models.generate_content(
        model=MODEL,
        contents=prompt
    )

    return response.text


print("Federal Urdu University AI Agent started.")
print("Ask me about teachers, subjects, or general university information.")
print("Type 'exit' to stop.\n")


while True:

    user = input("You: ")

    if user.lower() == "exit":
        print("Goodbye!")
        break

    try:
        answer = university_agent(user)
        print("\nAgent:", answer, "\n")

    except Exception as e:
        print("Error:", e)

```
