```python
budget = int(input("Enter your budget: "))

flights = {
    "F101": 250,
    "F102": 180,
    "F103": 220
}

suitable = {
    flight: price
    for flight, price in flights.items()
    if price <= budget
}

if suitable:
    flight = max(suitable, key=suitable.get)
    print("Recommended Flight:", flight)
    print("Price: $", suitable[flight])
else:
    print("No flight available within your budget.")
```
