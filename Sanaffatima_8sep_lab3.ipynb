```python
!pip install -U google-genai

from google import genai
Client = genai.Client(api_key="")

MODEL = "models/gemini-3.5-flash"


def notes_maker(user_input):

    prompt = f"""
You are an efficient lecture-wise notes maker.

You will get input as:
1) Teacher's recorded audio lecture
2) Board pictures of numericals or points taken in the class
3) Slides and notes provided by teacher
4) Related information from external sources if required

Make notes according to all the given input with practice problems.

User request: {user_input}

Prepared notes:
"""

    response = client.models.generate_content(
        model=MODEL,
        contents=prompt
    )

    return response.text


print("Gemini AI notes maker started (type 'exit' to stop)\n")

while True:
    user = input("You: ")

    if user.lower() == "exit":
        print("Goodbye")
        break

    try:
        reply = notes_maker(user)
        print("\nAgent:\n", reply, "\n")

    except Exception as e:
        print("Error:", e)
```
